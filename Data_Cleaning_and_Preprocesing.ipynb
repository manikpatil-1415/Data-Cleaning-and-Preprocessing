```python
import pandas as pd
import os

# Load all files
annex1 = pd.read_csv('/content/sales_data/annex1.csv')
annex2 = pd.read_csv('/content/sales_data/annex2.csv')
annex3 = pd.read_csv('/content/sales_data/annex3.csv')
annex4 = pd.read_csv('/content/sales_data/annex4.csv')

# Clean column names
annex1.columns = ['item_code', 'item_name', 'category_code', 'category_name']

annex2.columns = [
    'date',
    'time',
    'item_code',
    'quantity_sold_kg',
    'unit_selling_price_rmb_per_kg',
    'sale_or_return',
    'discount'
]

annex3.columns = [
    'date',
    'item_code',
    'wholesale_price_rmb_per_kg'
]

annex4.columns = [
    'item_code',
    'item_name',
    'loss_rate_percent'
]

# Convert dates
annex2['date'] = pd.to_datetime(annex2['date'], errors='coerce')
annex3['date'] = pd.to_datetime(annex3['date'], errors='coerce')

# Convert time
annex2['time'] = pd.to_datetime(
    annex2['time'],
    format='%H:%M:%S.%f',
    errors='coerce'
).dt.time

# Remove exact duplicate rows
annex1 = annex1.drop_duplicates()
annex2 = annex2.drop_duplicates()
annex3 = annex3.drop_duplicates()
annex4 = annex4.drop_duplicates()

print("Cleaning completed successfully!")
print("annex1:", annex1.shape)
print("annex2:", annex2.shape)
print("annex3:", annex3.shape)
print("annex4:", annex4.shape)
```

```python
# Check missing values after cleaning

cleaned_data = {
    "annex1": annex1,
    "annex2": annex2,
    "annex3": annex3,
    "annex4": annex4
}

for name, data in cleaned_data.items():
    print("\n==============================")
    print("File:", name)
    print("Missing values after cleaning:")
    print(data.isnull().sum())
```

```python
# Reload original annex2 file
raw_annex2 = pd.read_csv('/content/sales_data/annex2.csv')

# Find time values that could not be converted
converted_time = pd.to_datetime(
    raw_annex2['Time'],
    format='%H:%M:%S.%f',
    errors='coerce'
)

invalid_times = raw_annex2.loc[
    converted_time.isna(),
    ['Date', 'Time']
]

print("Invalid time values:", len(invalid_times))
display(invalid_times.head(20))
```

```python
# Reload original annex2
annex2 = pd.read_csv('/content/sales_data/annex2.csv')

# Clean column names
annex2.columns = [
    'date',
    'time',
    'item_code',
    'quantity_sold_kg',
    'unit_selling_price_rmb_per_kg',
    'sale_or_return',
    'discount'
]

# Convert date
annex2['date'] = pd.to_datetime(
    annex2['date'],
    errors='coerce'
)

# Handle both time formats
annex2['time'] = pd.to_datetime(
    annex2['time'],
    format='mixed',
    errors='coerce'
).dt.time

# Remove duplicate rows
annex2 = annex2.drop_duplicates()

print("Missing values after corrected cleaning:")
print(annex2.isnull().sum())
```

```python
# Check numerical columns for possible outliers

print("Annex2 numerical columns summary:")
print(annex2[['quantity_sold_kg',
              'unit_selling_price_rmb_per_kg']].describe())
```

```python
# Check negative quantities
negative_qty = annex2[annex2['quantity_sold_kg'] < 0]

print("Negative quantity records:", len(negative_qty))

print("\nSale/Return values for negative quantities:")
print(negative_qty['sale_or_return'].value_counts())

print("\nSample negative quantity records:")
display(
    negative_qty[
        ['date', 'time', 'item_code',
         'quantity_sold_kg', 'sale_or_return']
    ].head(10)
)
```

```python
print("Sale/Return unique values:")
print(annex2['sale_or_return'].value_counts())

print("\nDiscount unique values:")
print(annex2['discount'].value_counts())
```

```python
# Create folder for cleaned datasets
os.makedirs('/content/cleaned_data', exist_ok=True)

# Save cleaned files
annex1.to_csv('/content/cleaned_data/annex1_cleaned.csv', index=False)
annex2.to_csv('/content/cleaned_data/annex2_cleaned.csv', index=False)
annex3.to_csv('/content/cleaned_data/annex3_cleaned.csv', index=False)
annex4.to_csv('/content/cleaned_data/annex4_cleaned.csv', index=False)

print("All cleaned files saved successfully!")

print(os.listdir('/content/cleaned_data'))
```
